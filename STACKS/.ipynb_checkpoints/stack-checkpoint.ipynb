**STACK - LIFO DATA STRUCTURE**

LAST IN FIRST OUT i.e LAST ELEMENT IS ADDED IS FIRST ELEMENT TO RETRIEVE

PUSH - PUSH THE ITEM TO TOP OF THE STACK

POP - REMOVE THE TOP ITEM

PEEK - SEE TOP ITEM

**STACK APPLICATIONS**

REVERSE POLISH NOTATION FOR EVALUATING ARITHMETIC NOTATIONS

USED IN RECURSION

UNDO & REDO IN WORD PROCESSOR

```python
class Stack:
    def __init__(self):
        self.items = []

    def isEmpty(self):
        return not self.items

    def push(self,item):
        self.items.append(item)

    def pop(self):
        return self.items.pop()

    def peek(self):
        return self.items[-1]

    def size(self):
        return len(self.items)

    def __str__(self):
        return str(self.items) 


if __name__ == "__main__":
    s = Stack()
    print(s.isEmpty())
    s.push(3)
    print(s)
    s.push(7)
    s.push(9)
    print(s)
    print(s.pop())
    print(s)   
    print(s.peek())
    print(s.size())
```
